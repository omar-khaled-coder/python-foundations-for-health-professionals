## Brief

You receive these admissions on a paper sheet:

| Patient ID | Age | Gender | Ward             | LOS expected (days) |
|-----------|-----|-----|------------------|---------------------|
| P0101     | 78  | F   | Geriatrics       | 6                   |
| P0102     | 54  | M   | Cardiology       | 3                   |
| P0103     | 71  | F   | InternalMedicine | 5                   |
| P0104     | 39  | M   | Surgery          | 2                   |
| P0105     | 83  | F   | Geriatrics       | 9                   |

## Tasks
1. **Build** a list of 5 dictionaries called `roster` — one dict per patient, with the five fields above (use sensible key names: `patient_id`, `age`, `gender`, `ward`, `expected_los`).
2. **Filter:** create a new list `elderly` containing only patients aged 65 or older.
3. **Aggregate:** print the **count** of elderly patients and their **mean age**, formatted to one decimal.

> **Fast-Track (if you finish early):** Sort the original roster by `expected_los` *descending* — print the resulting list with one patient per line, formatted as `P0103 | LOS 5 | InternalMedicine`.

## Step 1 — Build the roster

```python
# TODO: build a list of 5 dicts called `roster`.
# Each dict should have keys: patient_id, age, gender, ward, expected_los

roster = [
    {"patient_id": "P0101", "age": 78, "gender": "F", "ward": "Geriatrics", "expected_los": 6},
    {"patient_id": "P0102", "age": 54, "gender": "M", "ward": "Cardiology", "expected_los": 3},
    {"patient_id": "P0103", "age": 71, "gender": "F", "ward": "InternalMedicine", "expected_los": 5},
    {"patient_id": "P0104", "age": 39, "gender": "M", "ward": "Surgery", "expected_los": 2},
    {"patient_id": "P0105", "age": 83, "gender": "F", "ward": "Geriatrics", "expected_los": 9}
]

# sanity check
print(len(roster), "patients in roster")
print(roster[0] if roster else "still empty")
```

## Step 2 — Filter for patients aged 65+

```python
# TODO: build a new list `elderly` with only patients where age >= 65.
elderly = []  # replace this
for p in roster:
    # We check the "age" key in the current patient dictionary
    if p["age"] >= 65:
        elderly.append(p)
print(len(elderly), "elderly patients")
```

## Step 3 — Count and mean age

```python
# TODO: print:
#   1. how many elderly patients there are
#   2. their mean age, formatted to one decimal place

# Hint: sum() and len() on a list of ages will get you there.

# print(f"Elderly count: ...")
# print(f"Mean age: ...")
# 1. Pull out just the ages into a simple list of numbers
elderly_ages = [p["age"] for p in elderly]

# 2. Calculate the count and the mean
count = len(elderly_ages)

if count > 0:
    mean_age = sum(elderly_ages) / count

    # 3. Print the results using f-strings
    print(f"Elderly count: {count}")
    print(f"Mean age: {mean_age:.1f}")
else:
    print("No elderly patients found.")
```

## ⭐ Fast-Track (only if you have time left)

```python
# TODO: sort `roster` by expected_los descending.
# Then print one line per patient like:
#   P0103 | LOS 5 | InternalMedicine
# 1. Sort the roster.
# 'key' tells Python to look at 'expected_los'.
# 'reverse=True' makes it descending (highest number first).
roster.sort(key=lambda p: p["expected_los"], reverse=True)

# 2. Print one line per patient with specific formatting
for p in roster:
    print(f"{p['patient_id']} | LOS {p['expected_los']} | {p['ward']}")
```
